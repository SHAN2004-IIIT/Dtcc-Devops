```python
import pandas as pd
import numpy as np

df = pd.read_csv("ecommerce_missing_duplicates_50_records.csv")
```

```python
#Identify and count missing values
df.isnull().sum()
```

```python
#Find repeated orders
df[df.duplicated(subset=["Order_ID"], keep=False)]
```

```python
df["Order_ID"].duplicated().sum()
```

```python
#Identify the highest-priced product
df.loc[df["Unit_Price"].idxmax(), ["Product", "Unit_Price"]]
```

```python
#Calculate total revenue grouped by city
df["Revenue"] = df["Unit_Price"] * df["Quantity"]

df.groupby("City")["Revenue"].sum().sort_values(ascending=False)
```
